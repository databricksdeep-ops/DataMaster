```python
# `first()[0]` — a common pattern for pulling a single scalar out of a Spark DataFrame
# `first()` returns the first Row; `[0]` indexes the first column of that Row.

data = [(1, "apple"), (2, "banana"), (3, "cherry")]
df = spark.createDataFrame(data, ["id", "fruit"])

# display(df)

# Grab the value of a single column as a plain Python scalar
first_id = df.select("id").first()
first_fruit = df.select("fruit").first()

print(first_id, first_fruit)  # 1 apple

# # Works on filtered/aggregated one-row results too
# max_id = df.agg({"id": "max"}).first()[0]
# print(max_id)  # 3

# # Equivalent alternatives: collect()[0][0], or df.first()["id"] (named-column access)
# print(df.select("id").collect()[0][0])  # 1
# print(df.first()["fruit"])              # apple
```
